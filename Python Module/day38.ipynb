```python
import time
#let us create two lists with around 1000000 elements
l1 = list(range(1000000))
l2 = list(range(1000000))
start = time.time()
l3 = [x+y for x,y in zip(l1,l2)]
end = time.time()
print(f"Execution time for lissts is {(end - start)*1000} milliseconds")
```

```python
#create elements in numpy arrays usse --> arange()
A1 = np.arange(1000000)
A2 = np.arange(1000000)
start = time.time()
A3 = A1 + A2
end = time.time()
print(f'execution time for numpy arrays is {(end - start) * 1000} milliseconds')
```
